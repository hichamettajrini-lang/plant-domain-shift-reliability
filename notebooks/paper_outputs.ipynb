```python
import os
import shutil

if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

os.makedirs("paper_outputs", exist_ok=True)

files_to_copy = [
    "results/final_20000_complete_summary.csv",
    "results/final_20000_performance_uncertainty_summary.csv",
    "results/final_20000_decision_errors.csv",
    "results/final_20000_threshold_tuning_results.csv",
    "results/final_20000_best_thresholds.csv",
    "results/comparison_20000_vs_30000_metrics.csv",
    "results/final_20000_performance_uncertainty_plot.png",
    "results/final_20000_decision_cost_plot.png",
    "results/final_20000_threshold_tuning_decision_cost.png",
    "results/comparison_20000_vs_30000_f1_score.png",
    "results/global_dataset_summary.csv",
    "results/global_dataset_summary.png"
]

for file_path in files_to_copy:
    if os.path.exists(file_path):
        shutil.copy(file_path, "paper_outputs")
        print("Copied:", file_path)
    else:
        print("Missing:", file_path)

print("\nFinished copying files to paper_outputs.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

metrics_2000 = pd.read_csv("results/mobilenetv3_small_all_dataset_metrics_comparison.csv")
metrics_20000 = pd.read_csv("results/mobilenetv3_small_plantvillage_20000_all_metrics.csv")
metrics_30000 = pd.read_csv("results/mobilenetv3_small_plantvillage_30000_all_metrics_epoch3.csv")

metrics_2000["training_sample_size"] = 2000
metrics_2000["saved_epoch"] = 3

if "saved_epoch" not in metrics_20000.columns:
    metrics_20000["saved_epoch"] = 3

all_size_comparison = pd.concat(
    [metrics_2000, metrics_20000, metrics_30000],
    ignore_index=True
)

all_size_comparison = all_size_comparison[
    [
        "train_dataset",
        "test_dataset",
        "model",
        "training_sample_size",
        "saved_epoch",
        "accuracy",
        "precision",
        "recall",
        "f1_score",
        "roc_auc"
    ]
]

output_path = "results/comparison_2000_20000_30000_metrics.csv"
all_size_comparison.to_csv(output_path, index=False)

print("Saved at:", output_path)
print(all_size_comparison)
```
