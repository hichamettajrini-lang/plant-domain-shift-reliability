```python
import os
import shutil
import pandas as pd
import matplotlib.pyplot as plt

if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

figures_dir = "paper_outputs/figures"
os.makedirs(figures_dir, exist_ok=True)

print("Project root:", os.getcwd())
print("Figures folder:", figures_dir)
```

```python
from matplotlib.patches import Rectangle

fig, ax = plt.subplots(figsize=(10, 4))
ax.axis("off")

boxes = [
    (0.05, 0.55, "PlantVillage\nTraining domain"),
    (0.28, 0.55, "MobileNetV3-Small\nLightweight model"),
    (0.52, 0.70, "PlantVillage\nValidation"),
    (0.52, 0.45, "PlantDoc\nExternal target"),
    (0.52, 0.20, "Plant Pathology 2020\nExternal target"),
    (0.78, 0.45, "Performance + Uncertainty\n+ Decision metrics")
]

for x, y, text in boxes:
    ax.add_patch(Rectangle((x, y), 0.18, 0.15, fill=False))
    ax.text(x + 0.09, y + 0.075, text, ha="center", va="center", fontsize=9)

arrows = [
    ((0.23, 0.625), (0.28, 0.625)),
    ((0.46, 0.625), (0.52, 0.775)),
    ((0.46, 0.625), (0.52, 0.525)),
    ((0.46, 0.625), (0.52, 0.275)),
    ((0.70, 0.775), (0.78, 0.525)),
    ((0.70, 0.525), (0.78, 0.525)),
    ((0.70, 0.275), (0.78, 0.525)),
]

for start, end in arrows:
    ax.annotate("", xy=end, xytext=start, arrowprops=dict(arrowstyle="->"))

plt.title("Cross-Dataset, Uncertainty-Aware, and Decision-Centric Evaluation Workflow")
plt.tight_layout()

figure1_path = "paper_outputs/figures/Figure_1_workflow.png"
plt.savefig(figure1_path, dpi=300)
plt.show()

print("Saved:", figure1_path)
```

```python
comparison_csv = "results/comparison_2000_20000_30000_metrics.csv"

comparison_df = pd.read_csv(comparison_csv)

plot_df = comparison_df.pivot_table(
    index="test_dataset",
    columns="training_sample_size",
    values="f1_score"
)

plot_df.plot(kind="bar", figsize=(9, 5))

plt.title("Effect of Training Sample Size on Cross-Dataset F1-score")
plt.xlabel("Test Dataset")
plt.ylabel("F1-score")
plt.ylim(0, 1.05)
plt.xticks(rotation=20)
plt.tight_layout()

figure3_path = "paper_outputs/figures/Figure_3_training_sample_size_f1_score.png"

plt.savefig(figure3_path, dpi=300)
plt.show()

print("Saved:", figure3_path)
```
